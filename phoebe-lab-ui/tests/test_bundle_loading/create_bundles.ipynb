```python
import phoebe
```

```python
# create bundles for each morphology type without UI-specific parameters:
bundle_detached = phoebe.default_binary()
bundle_semi_detached = phoebe.default_binary(semidetached='secondary')
bundle_contact = phoebe.default_binary(contact_binary=True)

bundles = [bundle_detached, bundle_semi_detached, bundle_contact]
```

```python
# add a dataset to each bundle:
for b in bundles:
    b.add_dataset(
        'lc',
        compute_times=phoebe.linspace(-0.5, 0.5, 201),
        times=[0.0, 0.1, 0.2],
        fluxes=[1.0, 0.9, 1.0],
        sigmas=[0.1, 0.1, 0.1],
        passband='Johnson:V'
    )
```

```python
# save each bundle:
bundle_detached.save('detached.bundle')
bundle_semi_detached.save('semi_detached.bundle')
bundle_contact.save('contact.bundle')
```

```python
# now let's add UI-specific parameters to each bundle and save again:
from phoebe.parameters import ChoiceParameter, ParameterSet

phoebe.parameters.parameters._contexts.append('ui')

parameters = []
parameters += [ChoiceParameter(
    qualifier='morphology',
    value='Detached',
    choices=['Detached', 'Semi-detached', 'Contact'],
    description='Morphology of the binary system',
    context='ui',
    component='binary'
)]

pset = ParameterSet(parameters)
for b in bundles:
    b._attach_params(pset)
```

```python
# set parameters and save the bundles:
bundle_detached.set_value('morphology', 'Detached', context='ui')
bundle_detached.save('detached_ui.bundle')

bundle_semi_detached.set_value('morphology', 'Semi-detached', context='ui')
bundle_semi_detached.save('semi_detached_ui.bundle')

bundle_contact.set_value('morphology', 'Contact', context='ui')
bundle_contact.save('contact_ui.bundle')
```
